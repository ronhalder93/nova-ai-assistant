# Nova Brain — train your own tiny language model

This notebook trains the small neural model in `nova-brain/` from random initialization. It does **not** call a hosted chat AI. This starter model is educational and will not be comparable to ChatGPT; quality depends heavily on the text corpus and compute.

## Steps
1. Run the setup cell.
2. Upload a UTF-8 plain-text file named `data.txt` containing text you own or are licensed to use (at least 200 characters; a larger, clean corpus is better).
3. Run training, then try generation.
4. Download `nova_tiny_model.pt` to keep your trained checkpoint.


```python
from pathlib import Path
import os, sys, subprocess
REPO = 'https://github.com/ronhalder93/nova-ai-assistant.git'
ROOT = Path('/content/nova-ai-assistant')
if not ROOT.exists():
    subprocess.run(['git','clone','--branch','nova-cloud-free','--depth','1',REPO,str(ROOT)], check=True)
BRAIN = ROOT / 'nova-brain'
os.chdir(BRAIN)
print('Working directory:', Path.cwd())
print('PyTorch available:', end=' ')
try:
    import torch
    print(torch.__version__)
except ImportError:
    subprocess.run([sys.executable,'-m','pip','install','-q','-r','requirements.txt'], check=True)
    import torch
    print(torch.__version__)
```

```python
from google.colab import files
uploaded = files.upload()  # Choose your own text file and name it data.txt
if 'data.txt' not in uploaded:
    raise ValueError('Please upload a plain-text file named exactly data.txt')
text_data = Path('data.txt').read_text(encoding='utf-8')
if len(text_data) < 200:
    raise ValueError(f'data.txt has only {len(text_data)} characters; add at least 200 characters.')
print(f'Corpus loaded: {len(text_data):,} characters')
```

```python
# Train the model from scratch. Default training may take a while on a free runtime.
subprocess.run([sys.executable, 'train.py'], check=True)
assert Path('nova_tiny_model.pt').exists(), 'Training did not create the model checkpoint.'
print('Training finished; checkpoint exists.')
```

```python
# Generate a short sample from the trained model.
prompt = input('Enter a short prompt: ')
subprocess.run([sys.executable, 'generate.py', '--prompt', prompt], check=True)
```
